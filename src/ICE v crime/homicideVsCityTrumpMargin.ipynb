# 2024 Presidential Election Margin vs Homicide Rate by State

This notebook visualizes the relationship between homicide rates in major U.S. cities and the 2024 presidential election margin in their respective states.

## Key Features:
- **X-axis**: Homicide rates per 100,000 (2024 data)
- **Y-axis**: Election margin (negative = Harris won, positive = Trump won)
- **Color coding**: Red for Trump states, Blue for Harris states

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch
import matplotlib.lines as mlines
```

## Data Definition

We'll define the cities, their election margins, homicide rates, and color coding based on which candidate won the state.

```python
# Cities data
cities = [
    "Jackson, MS",
    "Birmingham, AL",
    "St. Louis, MO",
    "Memphis, TN",
    "New Orleans, LA",
    "Detroit, MI",
    "Baltimore, MD",
    "Indianapolis, IN",
    "Richmond, VA",
    "Columbus, OH",
    "Minneapolis, MN",
    "St. Paul, MN",
]

# X-axis: Election margin (negative = Harris won, positive = Trump won)
margins = [24.3, 25.4, 18.4, 29.7, 22.0, 1.4, -28.5, 18.9, -5.2, 11.2, -4.2, -4.2]

# Y-axis: Homicide rates per 100,000 (2024 data)
homicide_rates = [
    77.5,
    58.8,
    54.1,
    48.0,
    34.7,
    47.9,
    34.7,
    27.1,
    23.0,
    22.4,
    17.95,
    10.85,
]

# Colors: Red for Trump states, Blue for Harris states
colors = [
    "red",
    "red",
    "red",
    "red",
    "red",
    "red",
    "blue",
    "red",
    "blue",
    "red",
    "blue",
    "blue",
]
```

## Visualization

Now we'll create a scatter plot showing the relationship between homicide rates and election margins.

```python
# Create the scatter plot
fig, ax = plt.subplots(figsize=(12, 8))

# Plot points
for i in range(len(cities)):
    ax.scatter(
        homicide_rates[i],
        margins[i],
        c=colors[i],
        s=250,
        alpha=0.7,
        edgecolors="black",
        linewidth=1.5,
    )
    # Add city labels with custom offsets to avoid overlap
    city_name = cities[i].split(",")[0]
    if city_name == "Richmond":
        # Place Richmond label below the point
        xytext_offset = (5, -20)
        va = "top"
    elif city_name == "Minneapolis":
        # Place Minneapolis label above the point
        xytext_offset = (5, 10)
        va = "bottom"
    else:
        xytext_offset = (5, 5)
        va = "bottom"
    ax.annotate(
        city_name,
        (homicide_rates[i], margins[i]),
        xytext=xytext_offset,
        textcoords="offset points",
        fontsize=14,
        alpha=0.8,
        va=va,
    )

# Labels and title
ax.set_xlabel("Homicide Rate per 100,000 (2024 Data)", fontsize=16, fontweight="bold")
ax.set_ylabel(
    "2024 Election Margin (% - Harris -, Trump +)",
    fontsize=16,
    fontweight="bold",
)
ax.set_title(
    "2024 Presidential Election Margin vs Homicide Rate by State",
    fontsize=18,
    fontweight="bold",
    pad=20,
)

# Increase tick label size
ax.tick_params(axis='both', which='major', labelsize=16)

# Legend
legend_elements = [
    Patch(facecolor="red", edgecolor="black", label="Trump Won State"),
    Patch(facecolor="blue", edgecolor="black", label="Harris Won State"),
]
ax.legend(handles=legend_elements, loc="lower right", fontsize=14)

# Grid
ax.grid(True, alpha=0.3, linestyle=":")

# Set axis limits with some padding
ax.set_xlim(0, 85)
ax.set_ylim(-35, 35)

plt.tight_layout()
plt.show()
```

## Bar Chart: Homicide Rates by City

Now we'll create a bar chart showing homicide rates for each city, sorted from highest to lowest, with color coding based on which candidate won the state.

```python
# Create a list of tuples (city, homicide_rate, margin, color)
data = list(zip(cities, homicide_rates, margins, colors))

# Sort by homicide rate (descending)
data_sorted = sorted(data, key=lambda x: x[1], reverse=True)

# Unpack sorted data
cities_sorted = [item[0] for item in data_sorted]
homicide_rates_sorted = [item[1] for item in data_sorted]
colors_sorted = [item[3] for item in data_sorted]

# Extract just city names (without state) for labels
city_labels = [city.split(",")[0] for city in cities_sorted]

# Create the bar chart
fig, ax = plt.subplots(figsize=(14, 8))

# Create bars
bars = ax.bar(range(len(cities_sorted)), homicide_rates_sorted, color=colors_sorted, 
             edgecolor='black', linewidth=1.5, alpha=0.8)

# Add city labels on x-axis with bold for Minneapolis and St. Paul
ax.set_xticks(range(len(cities_sorted)))
xtick_labels = []
for label in city_labels:
    if label in ['Minneapolis', 'St. Paul']:
        xtick_labels.append(label)
    else:
        xtick_labels.append(label)
ax.set_xticklabels(xtick_labels, rotation=45, ha='right', fontsize=14)

# Bold Minneapolis and St. Paul labels
for i, label in enumerate(xtick_labels):
    if label in ['Minneapolis', 'St. Paul']:
        ax.get_xticklabels()[i].set_fontweight('bold')

# Add value labels on top of each bar
for i, (bar, rate) in enumerate(zip(bars, homicide_rates_sorted)):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{rate}',
            ha='center', va='bottom', fontsize=12, fontweight='bold')

# Labels and title
ax.set_xlabel('City', fontsize=16, fontweight='bold')
ax.set_ylabel('Homicide Rate per 100,000 (2024 Data)', fontsize=16, fontweight='bold')
ax.set_title('Homicide Rates by City (Sorted Highest to Lowest)',
            fontsize=18, fontweight='bold', pad=20)

# Increase tick label size
ax.tick_params(axis='both', which='major', labelsize=14)

# Legend
legend_elements = [
    Patch(facecolor='red', edgecolor='black', label='Trump Won State'),
    Patch(facecolor='blue', edgecolor='black', label='Harris Won State'),
]
ax.legend(handles=legend_elements, loc='upper right', fontsize=14)

# Grid
ax.grid(True, alpha=0.3, linestyle=':', axis='y')

# Set y-axis limit with some padding
ax.set_ylim(0, max(homicide_rates_sorted) * 1.1)

plt.tight_layout()
plt.show()
```

## Analysis

This visualization allows us to observe patterns in homicide rates across cities with different political leanings. Key observations include:

- Cities in states that voted for Trump (red points) tend to have higher homicide rates

Note: This is a correlation analysis and does not imply causation. Many factors contribute to homicide rates beyond political affiliation.
